## 5.4 challenge -2 find the factors of number

```python
user_input=int(input("Enter the number"))
for i in range (1,user_input+1):
    if(user_input% i== 0):
        print(i)
```

```python
user_input=int(input("Enter the number"))
result=[]
for i in range (1,user_input+1):
    if(user_input% i== 0):
        result.append(i)
print("the factor of {} is {}".format(user_input,result))        
        
```
